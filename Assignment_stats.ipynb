#### Q26. Variance from a Frequency Distribution:- The following data represent the number of customers visiting a store.

> 

> **Calculate**:
> 1. Mean
> 2. Population variance
> 3. Population standard deviation

```python
import numpy as np

x_values = np.array([10, 15, 20, 25,30]) # उदा. मध्यबिंदू
f_values = np.array([3,5,8,6,3])     # उदा. वारंवारता

# १. Mean 
n = np.sum(f_values) 
print(n)
mean = np.sum(x_values * f_values) / n

# २. Population Variance 
variance = np.sum(f_values * (x_values - mean)**2) / n

# ३. Standard Deviation 
std_dev = np.sqrt(variance)

print(f"Mean: {mean:.2f}")
print(f"Population Variance: {variance:.2f}")
print(f"Population Standard Deviation: {std_dev:.2f}")
```

#### Q27. Detecting Outliers Using IQR:- The annual incomes (₹ thousands) of 12 employees are: 320, 340, 350, 360, 365, 370, 372, 375, 380, 385, 390, 520

> **Calculate**:
> 1. Q1
> 2. Q3
> 3. IQR
> 4. Lower and Upper Fences
> 5. Identify any outlier(s).

```python
import numpy as np

data = np.array([320, 340, 350, 360, 365, 370, 372, 375, 380, 385, 390, 520])

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1

lower_fence = q1 - (1.5 * iqr)
upper_fence = q3 + (1.5 * iqr)

outliers = data[(data < lower_fence) | (data > upper_fence)]

print(f"Q1: {q1}, Q3: {q3}, IQR: {iqr}")
print(f"Fences: {lower_fence} to {upper_fence}")
print(f"Outliers: {outliers}")
```

#### Q32. Two Coin Tosses:- Two fair coins are tossed simultaneously. Find the probability of obtaining:
> 1. Exactly one Head.
> 2. Two Heads.
> 3. At least one Head.

```python
outcomes = ['HH', 'HT', 'TH', 'TT']

# १. Exactly one Head
exactly_one_head = [x for x in outcomes if x.count('H') == 1]
# २. Two Heads
two_heads = [x for x in outcomes if x.count('H') == 2]
# ३. At least one Head
at_least_one_head = [x for x in outcomes if x.count('H') >= 1]

print(f"Exactly one head: {len(exactly_one_head)/len(outcomes)}")
print(f"Two heads: {len(two_heads)/len(outcomes)}")
print(f"At least one head: {len(at_least_one_head)/len(outcomes)}")
```

#### Q33. Three Coin Tosses:- Three fair coins are tossed. Find the probability of obtaining:
> 1. Exactly two Heads.
> 2. At most one Head.
> 3. No Heads.

```python
# Outcomes generate list
outcomes = []
coins = ['H', 'T']

# 3 coins ke liye nested loops
for c1 in coins:
    for c2 in coins:
        for c3 in coins:
            outcomes.append(c1 + c2 + c3)

total_outcomes = len(outcomes)

# Count function ka use karke calculation
def count_prob(condition_func):
    favorable = [o for o in outcomes if condition_func(o)]
    return len(favorable) / total_outcomes, len(favorable)

# 1. Exactly two Heads
prob2, count2 = count_prob(lambda o: o.count('H') == 2)

# 2. At most one Head
prob_at_most_one, count_at_most = count_prob(lambda o: o.count('H') <= 1)

# 3. No Heads
prob0, count0 = count_prob(lambda o: o.count('H') == 0)

# Results
print(f"Total Outcomes: {outcomes}")
print("-" * 30)
print(f"1. Exactly two heads: {prob2:.3f} ({count2}/8)")
print(f"2. At most one head: {prob_at_most_one:.3f} ({count_at_most}/8)")
print(f"3. No heads: {prob0:.3f} ({count0}/8)")
```

#### Q34. Rolling a Die:- A fair die is rolled once. Find the probability of obtaining:
> 1. An even number.
> 2. A number greater than 4.
> 3. A prime number.

```python
# Sample space of a die
die_outcomes = [1, 2, 3, 4, 5, 6]
total = len(die_outcomes)

# Conditions
even = [x for x in die_outcomes if x % 2 == 0]
greater_than_4 = [x for x in die_outcomes if x > 4]
prime = [x for x in die_outcomes if x in [2, 3, 5]]

print(f"1. Even probability: {len(even)}/{total} = {len(even)/total:.3f}")
print(f"2. > 4 probability: {len(greater_than_4)}/{total} = {len(greater_than_4)/total:.3f}")
print(f"3. Prime probability: {len(prime)}/{total} = {len(prime)/total:.3f}")
```

#### Q35. Two Dice:- Two fair dice are rolled simultaneously. Find the probability that:
> 1. The sum is 7.
> 2. The sum is greater than 9.
> 3. Both dice show the same number.

```python
# Sample space generate karna
outcomes = []
for d1 in range(1, 7):
    for d2 in range(1, 7):
        outcomes.append((d1, d2))

total_outcomes = len(outcomes)

# 1. Sum is 7
sum_7 = [o for o in outcomes if sum(o) == 7]

# 2. Sum is greater than 9
sum_gt_9 = [o for o in outcomes if sum(o) > 9]

# 3. Same number (Doubles)
same_num = [o for o in outcomes if o[0] == o[1]]

# Results display karna
print(f"Total Outcomes: {total_outcomes}")
print("-" * 30)
print(f"1. Probability sum is 7: {len(sum_7)}/{total_outcomes} = {len(sum_7)/total_outcomes:.3f}")
print(f"2. Probability sum > 9: {len(sum_gt_9)}/{total_outcomes} = {len(sum_gt_9)/total_outcomes:.3f}")
print(f"3. Probability same number: {len(same_num)}/{total_outcomes} = {len(same_num)/total_outcomes:.3f}")
```

#### Q43. In a class,
> - 60% of students study Python.
> - 45% study Statistics.
> - 30% study both.

> **Find**:
> 1. P(Python U Statistics)
> 2. P(Python | Statistics)

```python
def calculate_probability_stats(prob_a, prob_b, prob_intersection):
    """
    Calculates P(A U B) and P(A | B).
    
    Parameters:
    prob_a (float): Probability of event A
    prob_b (float): Probability of event B
    prob_intersection (float): Probability of both A and B (A ∩ B)
    """
    # 1. P(A U B) = P(A) + P(B) - P(A ∩ B)
    p_union = prob_a + prob_b - prob_intersection
    
    # 2. P(A | B) = P(A ∩ B) / P(B)
    # Check for division by zero
    if prob_b == 0:
        return p_union, "Undefined (P(B) is 0)"
    
    p_given = prob_intersection / prob_b
    
    return p_union, p_given

# --- Using the tool ---
# For our class problem:
# Python (A) = 0.60, Statistics (B) = 0.45, Both = 0.30
union, given = calculate_probability_stats(0.60, 0.45, 0.30)

print(f"P(Python U Statistics) = {union:.2f}")
print(f"P(Python | Statistics) = {given:.4f}")
```

#### Q47. A disease affects 2% of a population. A diagnostic test has:

#### Q79. A random variable X has the following distribution:
| X | 0 | 1 | 2 | 3 |
|---|---|---|---|---|
| P(X) | 0.20 | 0.35 | 0.30 | 0.15 |

> 1. E(X)
> 2. Var(X)
> 3. Standard deviation.

```python
x = [0, 1, 2, 3]
p = [0.20, 0.35, 0.30, 0.15]

# 1. Expected Value
e_x = sum(val * prob for val, prob in zip(x, p))

# 2. Variance
e_x2 = sum((val**2) * prob for val, prob in zip(x, p))
var_x = e_x2 - (e_x**2)

# 3. Standard Deviation
std_dev = var_x**0.5

print(f"E(X): {e_x:.2f}")
print(f"Var(X): {var_x:.2f}")
print(f"Standard Deviation: {std_dev:.4f}")
```

#### Q96. A manufacturer claims that the average weight of a packet is 500 g. A sample of 64 packets has:

> 1. State the null and alternative hypotheses.
> 2. Compute the Z statistic.
> 3. State the decision.
> 4. Draw the conclusion.

```python
import scipy.stats as stats
import math

# Given data
mu = 500
x_bar = 495
sigma = 20
n = 64
alpha = 0.05

# 1. Standard Error
se = sigma / math.sqrt(n)

# 2. Z statistic
z_stat = (x_bar - mu) / se

# 3. Critical value for two-tailed test at 5%
critical_z = stats.norm.ppf(1 - alpha/2)

print(f"Z-statistic: {z_stat}")
print(f"Critical Z: ±{critical_z:.2f}")

if abs(z_stat) > critical_z:
    print("Decision: Reject the Null Hypothesis")
else:
    print("Decision: Fail to reject the Null Hypothesis")
```
